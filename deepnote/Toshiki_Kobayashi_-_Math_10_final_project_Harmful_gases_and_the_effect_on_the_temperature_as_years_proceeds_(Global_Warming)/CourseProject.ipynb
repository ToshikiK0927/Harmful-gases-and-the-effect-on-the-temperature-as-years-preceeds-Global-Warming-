## Linear Regression
Here we will try to find the best fit a linear regression model using `CO2`(Carbon dioxide), `CH4`(Methane Gas), and `N2O`(Nitrogen gas) to predict `Temp` values

First import `LinearRegression` from sklearn and we instanciate it using `reg` as the vaiable name 

```
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
```

Now, we fit the model using `CO2`, `CH4`, `N2O` to predict `Temp` values

```
cols = ["CO2", "CH4", "N2O"]
reg.fit(df[cols], df["Temp"])
df["predict_temp"] = reg.predict(df[cols])
```

Lets check the accuracy of our prediction using Mean Squared Error (MSE) and Mean Absolute (MAE)

```
from sklearn.metrics import mean_squared_error
mean_squared_error(df["Temp"], df["predict_temp"])
```

```
from sklearn.metrics import mean_absolute_error
mean_absolute_error(df["Temp"], df["predict_temp"])
```

As you can see above, the linear regression model is pretty accurate due to the low errors obtained from both the MSE and MAE

Lets check the values of both the intercept and the coefficent to have an understanding of the relationship between `CO2`, `CH4`, `N2O` that predict `Temp` values

```
reg.intercept_
```

```
reg.coef_
```

The inetercept starts from negative since all of the data does not start from 0.
However, every coefficent in the model has a positive value meaning an increase in `CO2`, `CH4`, `N2O` has an increase in `Temp` values

We can visualize the graph of the relationship using altair between the actual relationship between `CO2`, `CH4`, `N2O` and `Temp` and compare it with `CO2`, `CH4`, `N2O` and `predict_temp` one column at a time 

```
import altair as alt
```

```
Carbon_chart = alt.Chart(df).mark_circle(color="red").encode(
    x = alt.X("CO2", scale = alt.Scale(zero=False)),
    y = alt.Y("Temp", scale = alt.Scale(zero=False))
)
Methane_chart = alt.Chart(df).mark_circle(color="blue").encode(
    x = alt.X("CH4", scale = alt.Scale(zero=False)),
    y = alt.Y("Temp", scale = alt.Scale(zero=False))
)
Nitrogen_chart = alt.Chart(df).mark_circle(color="green").encode(
    x = alt.X("N2O", scale = alt.Scale(zero=False)),
    y = alt.Y("Temp", scale = alt.Scale(zero=False))
)

Linear_Carbon_chart = alt.Chart(df).mark_circle(color="red").encode(
    x = alt.X("CO2", scale = alt.Scale(zero=False)),
    y = alt.Y("predict_temp", scale = alt.Scale(zero=False))
)
Linear_Methane_chart = alt.Chart(df).mark_circle(color="blue").encode(
    x = alt.X("CH4", scale = alt.Scale(zero=False)),
    y = alt.Y("predict_temp", scale = alt.Scale(zero=False))
)
Linear_Nitrogen_chart = alt.Chart(df).mark_circle(color="green").encode(
    x = alt.X("N2O", scale = alt.Scale(zero=False)),
    y = alt.Y("predict_temp", scale = alt.Scale(zero=False))
)

Carbon_chart & Linear_Carbon_chart| Methane_chart & Linear_Methane_chart|  Nitrogen_chart & Linear_Nitrogen_chart
```

## Polynomial Regression (machine learning)
Here we will try to find if as the year goes on, the temperature also rises due to the factors of the harmful gases. However, we will be training the data using train test split method for machine learning

Lets look at the graph to determine which degree to use

```
Year_chart = alt.Chart(df).mark_circle(color="red").encode(
    x = alt.X("Year", scale = alt.Scale(zero=False)),
    y = alt.Y("Temp", scale = alt.Scale(zero=False))
)
Year_chart
```

Looks like a degree of 3 will look good since the graph looks like it goes down at the start and starts increasing then decrases a bit at the end

First import `PolynomialFeatures` from sklearn and create the polynomial graph in the degree of 3

```
from sklearn.preprocessing import PolynomialFeatures
```

```
deg = 3
poly = PolynomialFeatures(degree = deg, include_bias = False)
year = pd.DataFrame(poly.fit_transform(df[["Year"]]))
year.columns = poly.get_feature_names_out()
```

Then, import the `train_test_split` from sklearn to train and test the data

```
from sklearn.model_selection import train_test_split
```

We are going to use the default train test split sizes which are 80% train and 20% test split. Then fit the data using `X_train` and `y_train`

```
seed = 1234
X_train, X_test, y_train, y_test = train_test_split(year, df["Temp"], random_state = seed)
reg = LinearRegression()
reg.fit(X_train, y_train)
```

Now, we are going to predict the data using X_test and X_train, and see its accuracy using Mean Squared Error

```
y_pred_train = reg.predict(X_train)
mean_squared_error(y_pred_train, y_train)
```

```
y_pred_test = reg.predict(X_test)
mean_squared_error(y_pred_test, y_test) 
```

The degree of 3 has a pretty accurate fit since the training error and especially the testing error has a very low value.

Now we are going to examine the intercepts and coefficents of the graph to see if the graph was overfitted or not

```
reg.intercept_
```

```
reg.coef_
```

The coef for index 1 and 2 above for each coefficent is pretty low. However, as we learned both of these values are being multiplied by the year raised to some power so, the low values of the coefficents makes sense. So, the graph was not overfitted.

Now, we are going to visualize and compare the predicted increase in temperature values with the actual increase temperature values as Years go on

```
import numpy as np
```

```
x_year = np.arange(min(df["Year"]), max(df["Year"]), 0.1)
df_poly = pd.DataFrame(
    {
        "x_year":x_year
    }
)
```

Here, we are calculating each predicted value of `y_pred_temp` using each value in `x_year` to show the relationship between year and predicted temperature

```
df_poly["y_pred_temp"] = reg.intercept_ 
for d in range(0,deg):
    df_poly["y_pred_temp"] += reg.coef_[d]*df_poly["x_year"]**(d+1)
```

```
poly_chart = alt.Chart(df_poly).mark_line().encode(
    x = alt.X("x_year", scale = alt.Scale(zero=False)),
    y = alt.Y("y_pred_temp", scale = alt.Scale(zero=False))
)
Year_chart + poly_chart
```

However the graph shows that a fairly polynomial relation with slight curves. It looks like we were sucessfully able to fit a polynomial graph by training our data by using degree of 3 polynomial without overfitting or underfitting the data

## K-Neighbors Regression (extra)
Lastly, we will run K-neighbors regression to predict the value of difference temperature using `CO2`, `CH4`, `N2O`

First, we train the data by performing a train test split using the default 80% train and 20% test split (remember `seed` is from the same random state number as the seed used for train test split in polynomial regression)

```
X_train, X_test, y_train, y_test = train_test_split(df[cols], df["Temp"], random_state = seed)
```

Now we import `KNeighborsRegressor` and instaciate it using `knn` with 10 neighbors
Then we fit the data using `X_train` and `y_train`

```
from sklearn.neighbors import KNeighborsRegressor
knn = KNeighborsRegressor(n_neighbors = 10)
knn.fit(X_train, y_train)
```

Lets predict the change in temperature and evalutate the error using MSE

```
df["K_predict_temp"] = knn.predict(df[cols])
mean_squared_error(df["Temp"], df["K_predict_temp"])
```

The error seems pretty low, which means we were sucessfully able to predict values by fitting `X_train` and `y_train`

Lets compare the actual graph with the predicted graph again to see how well the training data fitted

```
K_neighbor_Carbon_chart = alt.Chart(df).mark_circle(color="red").encode(
    x = alt.X("CO2", scale = alt.Scale(zero=False)),
    y = alt.Y("K_predict_temp", scale = alt.Scale(zero=False))
)
K_neighbor_Methane_chart = alt.Chart(df).mark_circle(color="blue").encode(
    x = alt.X("CH4", scale = alt.Scale(zero=False)),
    y = alt.Y("K_predict_temp", scale = alt.Scale(zero=False))
)
K_neighbor_Nitrogen_chart = alt.Chart(df).mark_circle(color="green").encode(
    x = alt.X("N2O", scale = alt.Scale(zero=False)),
    y = alt.Y("K_predict_temp", scale = alt.Scale(zero=False))
)

Carbon_chart & K_neighbor_Carbon_chart| Methane_chart & K_neighbor_Methane_chart| Nitrogen_chart & K_neighbor_Nitrogen_chart
```
